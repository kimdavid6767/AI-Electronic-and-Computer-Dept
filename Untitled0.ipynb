```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
# =========================
# 1. Define the dataset
# =========================
X = np.array([[1], [2], [3], [4], [5], [6]])  # Study hours
y = np.array([0, 0, 0, 1, 1, 1])             # Pass/Fail labels


# =========================
# 2. Create and train the logistic regression model
# =========================
model = LogisticRegression()
model.fit(X, y)

# Extract parameters (w, b)
w = model.coef_[0][0]
b = model.intercept_[0]
print(f"Logistic Regression function: f(x) = sigmoid({w:.4f} * x + {b:.4f})")
```

```python
# =========================
# 3. Generate x-values for the sigmoid curve
# =========================
x_range = np.linspace(0, 8, 200)        # Smooth curve
z = w * x_range + b
sigmoid = 1 / (1 + np.exp(-z))          # Sigmoid function
# =========================
# 4. Plot
# =========================
plt.figure(figsize=(8, 5))

# Scatter plot of actual data
plt.scatter(X, y, color='blue', label='Actual Data (Pass/Fail)', marker='o')

# Sigmoid curve
plt.plot(x_range, sigmoid, label='Sigmoid Curve', linewidth=2)

# Horizontal line at probability 0.5
plt.axhline(0.5, color='gray', linestyle='--', label='Decision Boundary (0.5)')

# Labels & Title
plt.xlabel("Study Hours (x)", fontsize=12)
plt.ylabel("Probability of Passing", fontsize=12)
plt.title("Logistic Regression Sigmoid Curve", fontsize=14)
plt.legend()
plt.grid(True)

plt.show()
```

```python
# =========================
# 5. Predict for x = 7
# =========================
x_new = np.array([[7]])
y_pred_prob = model.predict_proba(x_new)[0][1]
y_pred = model.predict(x_new)[0]

print(f"\nInput x = 7")
print(f"Probability of passing: {y_pred_prob:.4f}")
print(f"Predicted label: {y_pred}")
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LogisticRegression

# Length and height data for Dachshunds
dach_length = [55, 57, 64, 63, 58, 49, 54, 61]
dach_height = [30, 31, 36, 30, 33, 25, 37, 34]

# Length and height data for Jindo dogs
jin_length = [56, 47, 56, 46, 49, 53, 52, 48]
jin_height = [52, 52, 50, 53, 50, 53, 49, 54]

# Create labels: Dachshund = 0, Jindo = 1
X = np.array(
    [[l, h] for l, h in zip(dach_length + jin_length,
                          dach_height + jin_height)]
)
y = np.array([0] * len(dach_length) + [1] * len(jin_length))

# Print X and y
print("--- X (Feature data :: Matrix(2D array):16x2 ---")
print(X)

print("\n--- y (label data :: 1D Array(Vector)) ---")
print(y)

# Train logistic regression model
model = LogisticRegression()
model.fit(X, y)

# ---- Visualization ----
plt.scatter(dach_length, dach_height, c='r', label='Dachshund')
plt.scatter(jin_length, jin_height, c='b', marker='^', label='Jindo dog')

# --- (1) Predict model decision line ---
x_values = np.linspace(min(X[:, 0]) - 2, max(X[:, 0]) + 2, 200)

# decision boundary: w1*x + w2*y + b = 0  ->  y = -(w1*x + b)/w2
w1 = model.coef_[0][0]
w2 = model.coef_[0][1]
b = model.intercept_[0]

y_values = -(w1 * x_values + b) / w2

plt.plot(x_values, y_values, 'g--', linewidth=2, label='Decision Boundary')

# --- (2) New data come, let's predict ---
newdata_length = [59]
newdata_height = [35]

new_X = np.array([[newdata_length[0], newdata_height[0]]])

pred = model.predict(new_X)[0]
proba = model.predict_proba(new_X)[0]

names = {0: 'Dachshund', 1: 'Jindo dog'}
print(f"\nNew data: length={newdata_length[0]}, height={newdata_height[0]}")
print(f"Predicted class: {pred} ({names[pred]})")
print(f"Probability -> Dachshund: {proba[0]:.3f}, Jindo: {proba[1]:.3f}")

# Show the new point on the plot
plt.scatter(newdata_length, newdata_height, c='k', marker='*', s=250, label='New data')

# Keep the y-axis range readable (the boundary line can go far outside the data)
plt.ylim(min(X[:, 1]) - 3, max(X[:, 1]) + 3)

plt.xlabel('Length')
plt.ylabel('Height')
plt.title("Dog Classification with Decision Line Kim David 202390211")
plt.legend(loc='upper right')

plt.show()
```
