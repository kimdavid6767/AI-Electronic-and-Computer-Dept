```python
#######
import numpy as np
import matplotlib.pyplot as plt

# Data
x = [1, 3, 5, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [95, 80, 60, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]

# Create a regression model using a second-degree polynomial
mymodel = np.poly1d(np.polyfit(x, y, 2))

# Extract the coefficients in the form f(x) = ax^2 + bx + c
a, b, c = mymodel.coefficients

# Define the prediction function
def f(x_value):
    return mymodel(x_value)

# Print the prediction equation
print(f"f(x) = {a:.4f}x² + {b:.4f}x + {c:.4f}")

# Generate x-values for prediction
myline = np.linspace(1, 22, 100)

# Visualization
plt.scatter(x, y)
plt.plot(myline, f(myline), color='red')
plt.title("Polynomial Regression (degree=2)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
x = [1, 3, 5, 8, 9, 10, 12, 13, 14, 15, 16, 18, 19, 21, 22]
y = [95, 80, 60, 60, 65, 70, 70, 75, 76, 78, 79, 90, 99, 99, 100]

# Second-degree polynomial regression model
mymodel = np.poly1d(np.polyfit(x, y, 2))

# Coefficients: f(x) = ax^2 + bx + c
a, b, c = mymodel.coefficients

# Prediction function
def f(x_value):
    return mymodel(x_value)

print(f"f(x) = {a:.4f}x² + {b:.4f}x + {c:.4f}")

# ---- Prediction for 5:00 PM (17:00) ----
x_pred = 17
y_pred = f(x_pred)

# 1. Print the predicted speed
print(f"Predicted speed at {x_pred}:00 (5:00 PM) = {y_pred:.2f}")

# Generate x-values for the curve
myline = np.linspace(1, 22, 100)

# 2. Visualization with the predicted point
plt.scatter(x, y, label="Data")
plt.plot(myline, f(myline), color='red', label="Polynomial fit (degree 2)")

# Predicted point
plt.scatter(x_pred, y_pred, color='green', s=100, zorder=5,
            label=f"Prediction at 17:00 = {y_pred:.2f}")

# Dashed guide lines to the axes
plt.axvline(x_pred, color='green', linestyle='--', alpha=0.5)
plt.axhline(y_pred, color='green', linestyle='--', alpha=0.5)

# Label next to the point
plt.annotate(f"({x_pred}, {y_pred:.2f})",
             xy=(x_pred, y_pred),
             xytext=(x_pred - 5, y_pred + 5),
             arrowprops=dict(arrowstyle="->"))

plt.xlabel(" Kim David E40B Hour of the day")
plt.ylabel("Speed")
plt.title("Polynomial Regression (degree=2)")
plt.legend()
plt.show()
```
